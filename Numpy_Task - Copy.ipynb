```python
# 23.Check whether a matrix is symmetric or not

arr = np.array([
    [1,2,3],
    [2,4,5],
    [3,5,6]
])
symmetric = True
for i in range(len(arr)):
    for j in range(len(arr)):
        if arr[i][j] != arr[j][i]:
            symmetric = False
            break
if symmetric:
    print("Matrix is symmetric")
else:
    print("Metric is not symmetric")
```

```python
# 30.Find all duplicate elements in an array

array = np.array([10,20,30,20,10,50,10])

duplicat = []
for i in range(len(array)):
    for j in range(i + 1, len(array)):
        if array[i] == array[j] and array[i] not in duplicat:
            duplicat.append(array[i])
            
print("Duplicate element: ",duplicat)
```

```python
# 48. Find the longest increasing sequence in an array......
arr = [1, 2, 3, 2, 4, 5, 6]
longest = []
current = [arr[0]]
for i in range(1, (len(arr))):
    if arr[i] > arr[i-1]:
        current.append(arr[i])
    else:
        if len(current) > len(longest):
            longest = current
        current = [arr[i]]
print("longest:", longest)
```
